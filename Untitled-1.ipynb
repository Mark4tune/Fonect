```python
## Stemming ##
# Porter Stemming: English only
# Snowball Stemming: Improved porter, kadang akurat, support many languages
# Lancaster Stemming: Sangat cepat, kurang akurat, aggresive
```

```python
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer
import pandas as pd
```

```python
porter = PorterStemmer()
snowball = SnowballStemmer('english')
lancaster = LancasterStemmer()
```

```python
words = ["studies", "studying", "better", "children", "mice", "running", "happiness", "relational", "organization", "went", "universe", "university"]
```

```python
stemmed = []
for word in words:
    stemmed.append({
        "Original": word, 
        "Porter": porter.stem(word),
        "snowball": snowball.stem(word),
        "Lancaster": lancaster.stem(word)
    })

df = pd.DataFrame(stemmed)
print(df)
```

```python
from nltk.tag import pos_tag
from nltk.corpus import wordnet
```

```python
pos_tags = pos_tag(words)
print(pos_tags)
```

```python
def get_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("R"):
        return wordnet.ADV
    return wordnet.NOUN
```

```python
for word, tag in words:
    print(word, "-> ", lemmatizer.lematize(word, get_pos(tag)))
```

```python
## WordNet: lexical database, sets of synonyms (synonyms, antonyms, definition)
```

```python
from nltk.corpus import wordnet
```

```python
for word in words:
    print(word)
    synonyms = []
    antonyms = []
    
    synsets = wordnet.synsets(word)
    
    for syn in synsets:
        for lemma in syn.lemmas():
            synonyms.append(lemma.name())
            for ant in lemma.antonyms():
                antonyms.append(ant.name())
    
    print(f"Synonyms: {synonyms[:5]}")
    print(f"Antonyms: {antonyms}")
    
```

```python
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
sentence = "The children are playing better games and studying harder!!👌8===1"
```

```python
sentence = sentence.lower()
sentence = re.sub(r'\W', ' ', sentence)
sentence = re.sub(r'\d', ' ', sentence).strip()

tokens = word_tokenize(sentence)

stop_words = set(stopwords.words('english'))

words = []
for token in tokens:
    if token not in stop_words:
        words.append(token)
        
print(words)
```
